Write a program that finds the most common element in a list and how many times it appears.

```python
numbers=[20,30,2,5,56,12,3,2,4,4,4]
most_common = max(set(numbers), key=numbers.count)

print("Most common:", most_common)
print("Appears:", numbers.count(most_common), "times")
```

Create a program that counts the number of vowels and consonants in a sentence.

```python
sentence = input("Enter a sentence: ").lower()

vowels = "aeiou"
vowel_count = 0
consonant_count = 0

for char in sentence:
    if char.isalpha():
        if char in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

print("Vowels:", vowel_count)
print("Consonants:", consonant_count)
```

Write a program that finds the longest word in a sentence entered by the user.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

longest = max(words, key=len)

print("Longest word:", longest)
```

Create a program that sums user-entered numbers until the total exceeds 100, then stops and reports how many numbers were entered.

```python
total = 0
count = 0

while total <= 100:
    number = float(input("Enter a number: "))
    total += number
    count += 1

print("Total:", total)
print("Numbers entered:", count)
```

Create a program that prints the Fibonacci sequence up to a specific count provided by the user.


```python
count = int(input("How many Fibonacci numbers? "))

a = 0
b = 1

for i in range(count):
    print(a)
    a, b = b, a + b
```

Write a program that asks the user to enter numbers repeatedly and stops when the user enters a negative number, then prints the list of entered numbers.

```python
numbers = []

while True:
    number = int(input("Enter a number: "))

    if number < 0:
        break

    numbers.append(number)

print("Numbers entered:")
print(numbers)
```

Write a program that keeps asking for a number until the user enters 0, then outputs the sum of all entered numbers.

```python
total = 0

while True:
    number = float(input("Enter a number (0 to stop): "))

    if number == 0:
        break

    total += number

print("Sum:", total)
```

Create a program that finds the smallest power of 2 greater than a number entered by the user.

```python
number = int(input("Enter a number: "))

power = 1

while power <= number:
    power *= 2

print("Smallest power of 2 greater than", number, "is", power)
```
