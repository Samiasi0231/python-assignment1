#Write a program that asks the user for a radius and calculates both the area and circumference of a circle (use 3.14159 for π).

```python

radius = float(input("Enter the radius of the circle: "))
pi =float(input("Enter the value for pi: "))
area = pi * radius * radius
circumference = 2 * pi * radius
print(f"Area of the circle: {area}")
print(f"Circumference of the circle: {circumference}")
```

#Write a program that asks for two numbers and displays their sum, difference, product, quotient, remainder, and power (first number raised to the power of second number).

```python

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

sum_result = num1 + num2
difference = num1 - num2
product = num1 * num2

if num2 != 0:
    quotient = num1 / num2
    remainder = num1 % num2
else:
    quotient = "Undefined (cannot divide by zero)"
    remainder = "Undefined (cannot divide by zero)"

power = num1 ** num2

print(f"Sum: {sum_result}")
print(f"Difference: {difference}")
print(f"Product: {product}")
print(f"Quotient: {quotient}")
print(f"Remainder: {remainder}")
print(f"Power: {power}")
```

#Write a program that takes a user-entered temperature in Celsius and converts it to Fahrenheit, displaying the result.

#Create a program that calculates the Body Mass Index (BMI) by asking for a person's weight (in kg) and height (in meters).

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = weight / (height ** 2)

print(f"Your BMI is {bmi}")
```

#Create a program that calculates the simple interest for a principal amount, interest rate, and time period (in years) entered by the user.

```python
principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the interest rate (in %): "))
time = float(input("Enter the time (in years): "))
simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest
print(f"Simple Interest: {simple_interest}")
print(f"Total Amount: {total_amount}")
```

#Write a program that calculates the amount of paint needed to cover a wall, given its height and width in meters (assume 1 liter of paint covers 10 square meters).

#Create a program that takes a three-digit number and calculates the sum of its digits.


```python
number = int(input("Enter a three-digit number: "))

hundreds = number // 100
tens = (number // 10) % 10
units = number % 10

sum_digits = hundreds + tens + units

print(f"The sum of the digits is {sum_digits}")
```

#Write a program that calculates the hypotenuse of a right triangle based on the lengths of the other two sides entered by the user.

##Write a program that takes a base salary and calculates net pay after deducting taxes (assume a fixed tax rate of 20%).

```python

salary = float(input("Enter your base salary: "))

tax_rate = 20

tax = (tax_rate / 100) * salary
net_pay = salary - tax

print(f"Tax deducted: {tax}")
print(f"Net pay: {net_pay}")
```

##Write a program that takes a price without tax and calculates both the tax amount (at 7.5%) and the final price.

```python
price = float(input("Enter the price without tax: "))
tax_rate = 7.5
tax_amount = (tax_rate / 100) * price
final_price = price + tax_amount

print(f"Tax amount: {tax_amount}")
print(f"Final price: {final_price}")
```
