#Write a program that determines if a year entered by the user is a leap year. (Hint: A leap year is divisible by 4, except for century years which must be divisible by 400).

```python
year =int(input("enter a year:"))
if (year % 4 ==0 and year % 100 !=0) or (year % 400 ==0):
    print("Leap Year")
else:
    print("not a leap year")
    
```

#Create a simple calculator that takes two numbers and an operator (+, -, *, /) as input and performs the corresponding operation.

```python
a = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
b = float(input("Enter second number: "))

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b if b != 0 else "Cannot divide by zero")
else:
    print("Invalid operator")
```

#Write a program that categorizes a person by age: child (0-12), teenager (13-19), adult (20-64), or senior (65+).

```python
age = int(input("Enter age: "))

if 0 <= age <= 12:
    print("Child")
elif 13 <= age <= 19:
    print("Teenager")
elif 20 <= age <= 64:
    print("Adult")
else:
    print("Senior")
```

#Create a program that determines if a triangle is equilateral (all sides equal), isosceles (two sides equal), or scalene (no sides equal) based on three side lengths.

```python
a = float(input("Side 1: "))
b = float(input("Side 2: "))
c = float(input("Side 3: "))

if a == b == c:
    print("Equilateral")
elif a == b or b == c or a == c:
    print("Isosceles")
else:
    print("Scalene")
```

##Write a program that checks if a password meets basic criteria: at least 8 characters long and not equal to "password" or "123456".

```python
password = input("Enter password: ")

if len(password) >= 8 and password not in ["password", "123456"]:
    print("Valid password")
else:
    print("Invalid password")
```

##Write a program that checks if a string contains only digits, only letters, or a mix of both.

```python
text = input("Enter text: ")

if text.isdigit():
    print("Only digits")
elif text.isalpha():
    print("Only letters")
else:
    print("Mixed")
```

```python
#Create a program that determines if a word is a palindrome (reads the same forwards and backwards).
```

```python
word = input("Enter a word: ")

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
##Write a program that checks if a phone number is in the format "XXX-XXX-XXXX" where X is a digit.
```

```python
import re

phone = input("Enter phone number: ")

pattern = r"^\d{3}-\d{3}-\d{4}$"

if re.match(pattern, phone):
    print("Valid format")
else:
    print("Invalid format")
```
